```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the lab sample data
df = pd.read_csv("sample_lab_data.csv")

# Preview the data
df.head()

# --- TDS over Time ---
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="Date", y="TDS (mg/L)", hue="Location", marker="o")
plt.title("TDS (mg/L) Over Time by Location")
plt.xlabel("Date")
plt.ylabel("TDS (mg/L)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("charts/tds_over_time.png")
plt.show()

# --- Oil & Grease Boxplot ---
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x="Location", y="Oil & Grease (mg/L)", palette="Set2")
plt.title("Oil & Grease Concentration by Location")
plt.tight_layout()
plt.savefig("charts/oil_grease_boxplot.png")
plt.show()

# --- Basic Descriptive Statistics ---
df.describe()
```
